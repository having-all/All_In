# 변수
- 어떤 값을 저장하기 위한 메모리 공간
* 변수 선언 규칙
 - 변수 이름만 봐도 어떤 변수인지 해석할 수 있어야 함
 - 너무 함축적인 별명은 좋지 않음
 - 영문 대소문자, _, 숫자를 이용
 - 숫자로 시작하면 안 됨
 - 파이썬은 미리 변수를 선언하지 않아도 됨
 - 하지만 코드가 길면 변수를 미리 선언하는 것이 좋음
* 예약어(reserved keywords) 피하기
 - python에서 미리 선점한 단어들
 - 예시 : for, while, if, elif, else, class, try, except 등


```python
for = 10
print(for) # 에러 발생: 'for'는 예약어이므로 변수 이름으로 사용할 수 없습니다.
```

# 값 (Value)
- 값 할당
- 변수의 값이 변경되기 위해서는 =로 대입이 발생해야 함
- None : 아무 값을 갖지 않을 때 None 할당
- 일반적으로 변수를 생성할 때 사용
- 다른 언어에서는 null 등으로 사용됨

```python
temperature = 20
temperature - 10
temperature
```

```python
count, sum = None, None
print(count, sum)
```

# 연산자

```python
a = 10
b = 3
print(a+b)  # 더하기
print(a-b)  # 빼기
print(a*b)  # 곱하기
print(a/b) # 나누기
print(a//b) # 몫
print(a%b)  # 나머지
print(a**b) # 제곱
```
