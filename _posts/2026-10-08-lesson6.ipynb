# 함수
- 특정 기능을 하는 코드 블럭(묶음)
- 주어진 입력(input)에 대해 의도된 출력(output)을 전달하는 역할
- 반복적으로 호출해서 사용 가능
- 내장함수(built-in function): 파이썬 내부에 이미 정의(구현)되어 있는 함수
- 직접 필요에 따라 함수를 만들 수 있음

# print() 함수
- 값을 출력하는 함수
- ','로 여러 변수를 나열하면 같은 줄에 한 칸 띄어쓰기 후 출력
- print() 함수가 없어도 코드 마지막에 변수를 두면 값이 출력됨

```python
print('Hello World!')
print('Hello', 'World!')
a = 'Hello World!'
print(a)
```

# print() 함수 서식 지정
- %d : 정수(decimal)
- %f : 실수(float)
- %c : 문자 한 글자(character)
- %s : 한 글자 이상의 문자열(string)
